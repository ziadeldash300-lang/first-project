### Making new columns

```python
Route['Route name'] = Route['Departure Station'] + "-" + Route['Arrival Destination']
Route
```

```python
# Combined Master List of all unique stations
station_codes = {
    # London Hubs
    'London Euston': 'EUS',
    'London Paddington': 'PAD',
    'London Kings Cross': 'KGX',
    'London St Pancras': 'STP',
    'London Waterloo': 'WAT',

    # Major Cities
    'Liverpool Lime Street': 'LIV',
    'Manchester Piccadilly': 'MAN',
    'Birmingham New Street': 'BHM',
    'Bristol Temple Meads': 'BRI',
    'Edinburgh Waverley': 'EDB',
    'Edinburgh': 'EDB',
    'Glasgow Central': 'GLC',
    'Leeds': 'LDS',
    'Sheffield': 'SHF',
    'Newcastle': 'NCL',

    # Regional Stations
    'York': 'YRK',
    'Reading': 'RDG',
    'Oxford': 'OXF',
    'Durham': 'DHM',
    'Tamworth': 'TAM',
    'Wolverhampton': 'WVH',
    'Stafford': 'STA',
    'Doncaster': 'DON',
    'Swindon': 'SWI',
    'Nottingham': 'NOT',
    'Peterborough': 'PBO',
    'Crewe': 'CRE',
    'Leicester': 'LEI',
    'Nuneaton': 'NUN',
    'Didcot': 'DID',
    'Coventry': 'COV',
    'Wakefield': 'WKF',
    'Cardiff Central': 'CDF',
    'Warrington': 'WARR'
}


# This works for any column (Departure, Arrival, or a single Station column)
Route['Departure_Abbr'] = Route['Departure Station'].map(station_codes)
Route['Arrival_Abbr'] = Route['Arrival Destination'].map(station_codes)

# Now create combined "Route" column
Route['Route_Code'] = Route['Departure_Abbr'] + "-" + Route['Arrival_Abbr']
Route
```

```python
#season column
season_map = {
    12: 'Winter', 1: 'Winter', 2: 'Winter',
    3: 'Spring', 4: 'Spring', 5: 'Spring',
    6: 'Summer', 7: 'Summer', 8: 'Summer',
    9: 'Autumn', 10: 'Autumn', 11: 'Autumn'
}

Transaction['Season'] = Transaction['Date of Journey'].dt.month.map(season_map)
```

```python
hours = Transaction['Departure Time'].apply(lambda x: x.hour if pd.notnull(x) else None)

bins = [0, 5, 12, 17, 21, 24]
labels = ['Night', 'Morning', 'Afternoon', 'Evening', 'Night']

Transaction['Time Category'] = pd.cut(hours, bins=bins, labels=labels, include_lowest=True, ordered=False)
```

```python
Transaction['Day Type'] = Transaction['Date of Journey'].dt.dayofweek.apply(lambda x: 'Weekend' if x in [5, 6] else 'Weekday')
```

```python
import datetime as dt

# Function to combine date and time, handling NaT values
def combine_date_time_robust(date_obj, time_obj):
    if pd.isna(date_obj) or pd.isna(time_obj):
        return pd.NaT
    return pd.Timestamp.combine(date_obj.date(), time_obj)

sched_dt = Transaction.apply(lambda row: combine_date_time_robust(row['Date of Journey'], row['Arrival Time']), axis=1)
act_dt = Transaction.apply(lambda row: combine_date_time_robust(row['Date of Journey'], row['Actual Arrival Time']), axis=1)

is_next_day = act_dt < sched_dt
act_dt = act_dt.mask(is_next_day, act_dt + pd.Timedelta(days=1))

Transaction['Delay Duration'] = (act_dt - sched_dt).dt.total_seconds() / 60
```

```python
condition = (Transaction['Journey Status'] == 'On Time') | (Transaction['Delay Duration'] <= 0)

Transaction['Reason for Delay'] = np.select(
    condlist=[condition],
    choicelist=['No Delay'],
    default=Transaction['Reason for Delay']
)
Transaction['Refund Request'] = np.select(
    condlist=[condition],
    choicelist=['No'],
    default=Transaction['Refund Request']
)
```

```python
railcard_factor = Transaction['Railcard'].apply(lambda x: 1.0 if x == 'None' else 2/3)

ticket_type_map = {
    'Advance': 0.50,   # 1/2
    'Off-Peak': 0.75,  # 3/4
    'Anytime': 1.0     # Full Price
}
ticket_factor = Transaction['Ticket Type'].map(ticket_type_map).fillna(1.0)

Transaction['Original Ticket Price'] = Transaction['Price'] / (railcard_factor * ticket_factor)

Transaction['Original Ticket Price'] = Transaction['Original Ticket Price'].round(2)

print(Transaction[['Price', 'Railcard', 'Ticket Type', 'Original Ticket Price']].head())
```

### 2.Which ticket class (Standard vs. First Class) generates the highest revenue?

```python
revenue_by_ticket_class = df_analysis.groupby('Ticket Class')['Price'].sum().sort_values(ascending=False)

plt.figure(figsize=(8, 5))
sns.barplot(x=revenue_by_ticket_class.index, y=revenue_by_ticket_class.values, palette='viridis')
plt.title('Total Revenue by Ticket Class')
plt.xlabel('Ticket Class')
plt.ylabel('Total Revenue (£)')
plt.show()

print("Revenue by Ticket Class:\n", revenue_by_ticket_class)
```

#### Conclusion:
Standard Class tickets generate significantly more revenue than First Class tickets. This is expected as Standard Class typically has higher ridership volume, even if the per-ticket price is lower than First Class. This indicates that the bulk of the railway's revenue comes from mass-market offerings.

### 3.Which ticket type (Advance, Off-Peak, Anytime) contributes the most to total revenue?

#### Conclusion:
Advance tickets contribute the most to total revenue, followed by Off-Peak and Anytime. This suggests that passengers often plan their journeys in advance to take advantage of lower fares, or that the volume of Advance tickets sold is considerably higher. The pricing strategy for Advance tickets is crucial for overall revenue.

```python
df_analysis['Journey Month'] = df_analysis['Date of Journey'].dt.month_name()
df_analysis['Journey Hour'] = df_analysis['Departure Time'].apply(lambda x: x.hour)

monthly_delays = df_analysis[df_analysis['Journey Status'] == 'Delayed'].groupby('Journey Month').size().reindex(pd.to_datetime(['2024-01-01', '2024-02-01', '2024-03-01', '2024-04-01']).month_name(), fill_value=0)
monthly_cancellations = df_analysis[df_analysis['Journey Status'] == 'Cancelled'].groupby('Journey Month').size().reindex(pd.to_datetime(['2024-01-01', '2024-02-01', '2024-03-01', '2024-04-01']).month_name(), fill_value=0)

hourly_delays = df_analysis[df_analysis['Journey Status'] == 'Delayed'].groupby('Journey Hour').size().reindex(range(24), fill_value=0)
hourly_cancellations = df_analysis[df_analysis['Journey Status'] == 'Cancelled'].groupby('Journey Hour').size().reindex(range(24), fill_value=0)

plt.figure(figsize=(12, 6))
sns.barplot(x=monthly_delays.index, y=monthly_delays.values, palette='Greens')
plt.title('Monthly Delay Patterns')
plt.xlabel('Month')
plt.ylabel('Number of Delays')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()

plt.figure(figsize=(12, 6))
sns.barplot(x=monthly_cancellations.index, y=monthly_cancellations.values, palette='Blues')
plt.title('Monthly Cancellation Patterns')
plt.xlabel('Month')
plt.ylabel('Number of Cancellations')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()

plt.figure(figsize=(12, 6))
sns.barplot(x=hourly_delays.index, y=hourly_delays.values, palette='Reds')
plt.title('Hourly Delay Patterns')
plt.xlabel('Hour of Day')
plt.ylabel('Number of Delays')
plt.tight_layout()
plt.show()

plt.figure(figsize=(12, 6))
sns.barplot(x=hourly_cancellations.index, y=hourly_cancellations.values, palette='Purples')
plt.title('Hourly Cancellation Patterns')
plt.xlabel('Hour of Day')
plt.ylabel('Number of Cancellations')
plt.tight_layout()
plt.show()

print("Monthly Delays:\n", monthly_delays)
print("\nMonthly Cancellations:\n", monthly_cancellations)
print("\nHourly Delays:\n", hourly_delays)
print("\nHourly Cancellations:\n", hourly_cancellations)
```

### 6.How much revenue loss is specifically attributed to ticket type discounting (e.g., the price gap of Advance vs. Anytime tickets)?

```python
df_analysis['Revenue Loss from Ticket Type Discount'] = df_analysis['Original Ticket Price'] - df_analysis['Price']

revenue_loss_by_ticket_type = df_analysis.groupby('Ticket Type')['Revenue Loss from Ticket Type Discount'].sum().sort_values(ascending=False)

plt.figure(figsize=(10, 6))
sns.barplot(x=revenue_loss_by_ticket_type.index, y=revenue_loss_by_ticket_type.values, palette='cubehelix')
plt.title('Revenue Loss from Ticket Type Discounting')
plt.xlabel('Ticket Type')
plt.ylabel('Total Revenue Loss (£)')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()

print("Revenue Loss from Ticket Type Discounting:\n", revenue_loss_by_ticket_type)
```

#### Conclusion:
By comparing the `Original Ticket Price` with the `Price` paid, we can quantify the revenue loss attributed to different ticket type discounts. This analysis reveals that `Advance` tickets contribute the most to revenue loss through discounting, which is expected due to their lower price point designed to encourage early bookings. While these discounts attract passengers, understanding their cumulative impact is essential for optimizing pricing strategies to balance ridership and revenue generation.
